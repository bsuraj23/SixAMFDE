
Data and Configuration File Formats

.txt - Plain text files read and written using Python's built-in file handling functions..
csv - Comma-separated value files managed natively through the built-in csv module..
json - JavaScript Object Notation files parsed and generated using Python's standard json library..
pkl or .pickle - Binary files containing serialized Python objects, created using the pickle module.
.yaml or .yml - Human-readable data serialization files commonly used for configuration, parsed using external libraries like PyYAML.

Different data types used in Python

Python has several built-in data types used to store different kinds of values, categorized into numeric, sequence, mapping, set, boolean, binary, and null types. 

Numeric Types

int - Stores whole numbers like 5 or -3 without decimals.
float - Stores numbers with decimal points like 10.5 or 3.14.
complex - Stores numbers with real and imaginary parts like 3 + 5j.

Sequence Types

str - Stores text data enclosed in quotes like "Hello".
list - Stores an ordered, changeable collection of items like [1, 2, "apple"].
tuple - Stores an ordered, unchangeable collection of items like (1, 2, "apple").
range - Stores a sequence of numbers, often used in loops like range(0, 10).

Mapping Type

dict - Stores key-value pairs like {"name": "Alice", "age": 25}.
Set Typesset - Stores an unordered collection of unique items like {1, 2, 3}.
frozenset - Stores an immutable version of a set.

Boolean Type

bool - Stores truth values, which are either True or False.

Binary Types

bytes - Stores immutable byte data like b"Hello".
bytearray - Stores mutable byte data.
memoryview - Allows code to access internal data of an object that supports buffer protocols without copying.
