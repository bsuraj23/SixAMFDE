```python
# To reverse a string without using -1

text = "Python"

reversed_text = ""

for char in text:
    reversed_text = char + reversed_text

print(reversed_text)
```

```python
reversed_string=text[::-1]
print(reversed_string)
```

1. If there is no internet, how can we use an LLM to get code on our laptop?
Normally, online LLM services require internet. But you can use a local/offline LLM by downloading the model to your laptop beforehand.
The basic idea is:
Internet available initially → Download model → Store model locally → Run model on laptop → Use without internet
For example, tools such as Ollama can run supported open-source models locally. Once the required model files are downloaded, you can use the model without an internet connection.
