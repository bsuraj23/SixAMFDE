```python
#1.	Without using copy and [:], need to copy a list into a new one.

list1=[1,2,4,5]
list_copy=list(list1)
print(list_copy)
```

```python
#Using list comprehension

old_list = [1, 2, 3]
new_list = [item for item in old_list]
print(new_list)
```

```python

#Using .extend()

old_list = [1, 2, 3] 
new_list = [] 
new_list.extend(old_list)
print(new_list)
```

```python
# 3.	Reverse the elements in a list by skipping one element.

original_list = [1, 2, 3, 4, 5, 6, 7, 8]
reversed_list=original_list[::-1]
print(reversed_list)
```

```python
#Access the last character of a string without using -1, you can use len()

text = "Python"

last_char = text[len(text) - 1]

print(last_char)
```
